# Cluster only epithelial cells

```python
adata_epi=adata[adata.obs['cell class']=='Epithelial']
```

```python
del adata
```

```python
cnv.tl.pca(adata_epi)
cnv.pp.neighbors(adata_epi)
cnv.tl.leiden(adata_epi,resolution=0.2,key_added='cnv_epi')
sc.tl.dendrogram(adata_epi,groupby='cnv_epi')
```

```python
counts_clone=adata_epi.obs.groupby('cnv_epi').count()
counts_clone_norm=counts_clone.div(counts_clone.sum(axis=0),axis=1)*100
selected_clones=counts_clone_norm[counts_clone_norm['cell_id']>0.1].index
## criteria is subclones representing at least 0.1% of cells
```

```python
#filter cells
adata_epi=adata_epi[adata_epi.obs['cnv_epi'].isin(selected_clones)]
```
